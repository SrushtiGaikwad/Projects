```python
def rag_qa(question: str) -> str:
    """
    Runs full RAG pipeline:
    Retrieval → Augmentation → Generation
    """

    # 1. Retrieve relevant chunks
    retrieved_docs = retriever.invoke(question)

    if not retrieved_docs:
        return "I don't know. The answer is not present in the video."

    # 2. Build context
    context_text = "\n\n".join(doc.page_content for doc in retrieved_docs)

    # 3. Build prompt (augmentation)
    final_prompt = prompt.invoke({
        "context": context_text,
        "question": question
    })

    # 4. Generate answer
    response = model.generate_content(final_prompt.text)

    return response.text.strip()
```

Building a Chain

```python
from langchain_core.runnables import RunnableParallel, RunnablePassthrough, RunnableLambda
from langchain_core.output_parsers import StrOutputParser
```

```python
def format_docs(retrieved_docs):
  context_text = "\n\n".join(doc.page_content for doc in retrieved_docs)
  return context_text
```

```python
parallel_chain = RunnableParallel({
    'context': retriever | RunnableLambda(format_docs),
    'question': RunnablePassthrough()
})
```

```python
parallel_chain.invoke('what is Database')
```

```python
parser = StrOutputParser()
```

```python
main_chain = parallel_chain | prompt | llm | parser
```

```python
main_chain.invoke('Can you summarize the video')
```
